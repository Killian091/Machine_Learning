```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=100, n_features=2, n_informative=2, n_redundant=0, random_state=41,
                           n_classes=2, n_clusters_per_class=1, hypercube=False, class_sep=15)
y = np.where(y == 0, -1, 1) # Convert 0 to -1 and 1 to 1
```

```python
y.shape
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
```

```python
def perceptron(X,y):
  w1=w2=b=1
  lr=0.1

  for j in range(1000):

    for i in range(X.shape[0]):

      z = w1*X[i][0] + w2*X[i][1] + b

      if z*y[[i]] < 0:
        w1 = w1 + lr*y[i]*X[i][0]
        w2 = w2 + lr*y[i]*X[i][1]
        b = b + lr*y[i]

  return w1,w2,b
```

```python
w1,w2,b = perceptron(X,y)
```

```python
# Calculate slope and intercept of decision boundary
m = -(w1 / w2)
c = -(b / w2)

print(m, c)

# Generate points for the line
x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + c

# Plot
plt.figure(figsize=(10, 6))
plt.plot(x_input, y_input, color='red', linewidth=3)

# Scatter plot of data points
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100)

# Set y-axis limits
plt.ylim(-3, 2)

plt.show()
```
